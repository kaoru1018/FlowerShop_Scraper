```python
import requests
import re
import pandas as pd
from bs4 import BeautifulSoup
import time
from google.colab import files

# 1. 検索結果からURLを抜き出す関数
def get_flower_urls(query):
    search_url = f"https://search.yahoo.co.jp/search?p={query}"
    res = requests.get(search_url)
    # 検索結果からURLを抜き出す（正規表現）
    urls = re.findall(r'https://[\w/:%#\$&\?\(\)~\.=\+\-]+', res.text)
    # Yahoo関連のリンクを除外して、お花屋さんっぽいサイトに絞る（簡易フィルター）
    flower_urls = [u for u in list(set(urls)) if "yahoo" not in u and "google" not in u and "search" not in u]
    return flower_urls[:12] # 上位12件だけ取得

# 2. 情報を収集する関数
def scrape_flower_shop(url):
    try:
        res = requests.get(url, timeout=5)
        res.encoding = res.apparent_encoding
        soup = BeautifulSoup(res.text, 'html.parser')
        title = soup.title.string.strip() if soup.title else "No Title"
        desc = soup.find('meta', attrs={'name': 'description'})
        desc_text = desc['content'].strip() if desc and desc.has_attr('content') else "No Description"
        return [title, url, desc_text]
    except:
        return ["取得失敗", url, "取得不可"]

# --- メイン処理 ---
print("検索中...")
urls = get_flower_urls("お花屋さん")
data_list = []

for url in urls:
    print(f"解析中: {url}")
    time.sleep(2)
    data_list.append(scrape_flower_shop(url))

# 3. ファイル出力
df = pd.DataFrame(data_list, columns=['店名', 'URL', '説明文'])
df['強み/特徴'] = ""
df['調査メモ'] = ""

output_file = 'Auto_FlowerShops_List.csv'
df.to_csv(output_file, index=False, encoding='utf-8-sig')

print(f"\n完了！'{output_file}' をダウンロードします。")
files.download(output_file)
```

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
import time

# 12件のURLリスト
urls = [
    "https://www.i879.com",
    "https://hanazakari.jp",
    "https://mermer.jp",
    "https://www.nrc.co.jp",
    "https://hanaprime.jp",
    "https://hitohana.tokyo",
    "https://www.aoyamaflowermarket.com",
    "https://www.e87.com",
    "https://ohanano-madoguchi.com",
    "https://bloomeelife.com",
    "https://www.hibiyakadan.com",
    "https://www.eflora.co.jp"
]

print(f"合計 {len(urls)} 件のURLを処理します...")

data_list = []

# 処理開始
for url in urls:
    try:
        print(f"アクセス中: {url}")
        time.sleep(2) # サーバーへの負担を減らすため
        res = requests.get(url, timeout=10)

        # 文字化け対策（サイトの文字コードを自動判別する）
        res.encoding = res.apparent_encoding

        soup = BeautifulSoup(res.text, 'html.parser')

        # タイトル取得
        title = soup.title.string.strip() if soup.title else "No Title"

        # 説明文取得（metaタグのdescriptionを探す）
        desc = soup.find('meta', attrs={'name': 'description'})
        desc_text = desc['content'].strip() if desc and desc.has_attr('content') else "No Description"

        data_list.append([title, url, desc_text])
        print(" -> 成功！")
    except Exception as e:
        print(f" -> 失敗: {url}, エラー: {e}")

# データフレームを作成して保存
df_result = pd.DataFrame(data_list, columns=['店名', 'URL', '説明文'])

# utf-8-sig はExcelで開いた時に文字化けしにくい形式です
output_file = 'FlowerShops_Final_Result.csv'
df_result.to_csv(output_file, index=False, encoding='utf-8-sig')

print(f"\n完成しました！'{output_file}' をダウンロードします。")

# 自動でダウンロード
from google.colab import files
files.download(output_file)
```
